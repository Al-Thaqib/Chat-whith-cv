```python
model_name = "gpt-4o-mini"

system_prompt =  """Role: you are a Calculator router.
                 tasks: your Responsibility is to take the user prompt and decide his intent.
                you have 4 routes [*,+,-,/], you extract two numbers,[x,y].
                for Example user Prompt: add 2 to 3 route = *, x = 2, y = 3.
                for Example user Prompt: 3 * 5, route = *, x = 3, y = 5.
                output Json:{\"route:\"+\", \"x\":5, \"y\":7}."""
```

```python
def router(user_prompt):
    client = OpenAI()
    response = client.chat.completions.create(
        model = model_name,
        messages = [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ],
        response_format= {"type": "json_object"}
    )
    reply_from_ai = response.choices[0].message.content

    print(reply_from_ai)
```

```python
router("10 + 20")
```
