# 📄 Chat With CV — Interactive AI Career Assistant

## 📦 1. What libraries I need
Here we import all the tools required for our AI Assistant:
* `os`: To communicate with the operating system and get environment variables.
* `dotenv`: To securely read our secret API key from the `.env` file.
* `openai`: To communicate with the Large Language Model (LLM).
* `docx`: To read and extract text from our `.docx` resume file.

```python
# os to communicate with my system
import os
# dotenv to save my api key
from dotenv import load_dotenv
# openai -> OpenAI to chat with LLM
from openai import OpenAI
# docx to read the file 
import docx
```

## 🔑 2. Save and Verify My API-Key

We load the environment variables and check that the key exists before calling the AI.

```python
load_dotenv(override=True)
# Make sure that OPEN_AI_API_KEY is the name of api in .env
openai_api_key = os.getenv("OPEN_AI_API_KEY")
if openai_api_key:
    print(f"Your first four numbers of API-Key is {openai_api_key[:4]}")
else:
    print("Nothing")
```

## 📖 3. Get Text from my_cv.docx

We iterate through all paragraphs in `my_cv.docx`, collect them into a list, and join them with `\n` to make a single readable string.

```python
# I make list to append text on it to use func -> join that make it readable
cv_text = []
doc = docx.Document('my_cv.docx')
for page in doc.paragraphs:
    cv_text.append(page.text)
cv_readable = '\n'.join(cv_text)
print(cv_readable)
```

## 🤖 4. Talk with AI

### Notes on this step:
1. **`sytem_prompt`**: We inject `{cv_readable}` into the system instructions so the AI knows our career details.
2. **`base_url`**: We point to OpenRouter (`https://openrouter.ai/api/v1`) to route requests properly.
3. **`while True` Loop**: Lets the user ask multiple questions interactively. Type `exit` to stop.
4. **`response.choices[0].message.content`**: Extracts the generated text from the LLM response object.

```python
sytem_prompt = f"""
    you are my assistant. you are responsible to reply to questions about my career.
    you have my cv here:\n{cv_readable}. use it to reply
    """
model_name = 'openai/gpt-4o-mini'
# Setting OpenRouter base URL and passing the API key
LLM = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=openai_api_key
)
while True:
    user_prompt = input("whatis your question?\n")
    if user_prompt == 'exit':
        break
    messages = [
        {"role": "system", "content": sytem_prompt},
        {"role": "user", "content": user_prompt}
    ]
    response = LLM.chat.completions.create(
        model=model_name,
        messages=messages
    )
    # Extract the clean reply text
    reply_from_ai = response.choices[0].message.content
    print(reply_from_ai)
```
